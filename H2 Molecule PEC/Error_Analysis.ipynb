```python
import pandas as pd
import pickle

df_Aer = pd.read_pickle("Data_results/Aer_Simulator_Results/Aer_Results_2.pkl")

df_Svec = pd.read_pickle("Data_results/Statevector_Results/Statevector_Results.pkl")

df_combined = pd.read_pickle("Data_results/Combined_Results.pkl")
```

```python
path = "Noise_mitigation/"
df1 = pd.read_pickle(path + "Bond_45/Result_dataframe.pkl")
df2 = pd.read_pickle(path + "Bond_60/Result_dataframe.pkl")
df3 = pd.read_pickle(path + "Bond_75/Result_dataframe.pkl") 
df4 = pd.read_pickle(path + "Bond_90/Result_dataframe.pkl")
df5 = pd.read_pickle(path + "Bond_105/Result_dataframe.pkl")
df6 = pd.read_pickle(path + "Bond_120/Result_dataframe.pkl")

df_QPU_NoiseMitigation = pd.concat([df1, df2, df3, df4, df5, df6], ignore_index=True)
df_QPU_NoiseMitigation.head(10)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# data
bond = df_Aer["Bond Length"]
E_Aer = df_Aer["Ground State Energy (Hartree)"]
E_Svec = df_Svec["Ground State Energy (Hartree)"]

bond_small = df_combined["Bond Length"]
E_QPU_small = df_combined["Ground State Energy (Hartree)_QPU"]
E_Aer_small = df_combined["Ground State Energy (Hartree)_Aer"]
E_Svec_small = df_combined["Ground State Energy (Hartree)"]


# error
transpilation_error = np.abs(E_Svec - E_Aer)
hardware_noise = np.abs(E_Aer_small - E_QPU_small)
transpilation_error_small = np.abs(E_Svec_small - E_QPU_small)
total_error = np.abs(transpilation_error_small + hardware_noise)

# Create figure
fig, (ax1, ax2) = plt.subplots(
    2, 1,
    sharex=True,
    gridspec_kw={'height_ratios':[3,1]},
    figsize=(6,6)
)

# -------- Top energy plot --------
ax1.plot(bond_small, E_QPU_small, 'o--', label='QPU')
ax1.plot(bond, E_Aer, '.--', label='Aer Simulator')
ax1.plot(bond, E_Svec, 'x--', label='Statevector')



ax1.set_ylabel("Energy [Hartree]")
ax1.grid(True)
ax1.legend()

# -------- Bottom error plot --------
# ax2.scatter(bond_small, hardware_noise, marker='^', color="tab:cyan", label="Unaccounted Hardware Noise")
ax2.scatter(bond, transpilation_error, marker='.',color='tab:orange', label="Aer Simulator Noise")
ax2.scatter(bond_small, total_error, marker='o',color='tab:blue' ,label="Hardware Noise")


ax2.set_ylabel("E [Hartree]")
ax2.set_xlabel("Bond distance [Å]")
ax2.legend()

plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

bond_lengths = [0.45, 0.60, 0.75, 0.90, 1.05, 1.20]

E_sv  = [-0.997445, -1.116049, -1.135370, -1.089969, -1.056434, -1.056434]
E_aer = [-0.789255, -0.946851, -1.004726, -0.998906, -0.976473, -0.976473]
E_qpu = [-0.729711, -0.858650, -0.908009, -0.954234, -0.963867, -0.948036]
E_mit = [-0.967469, -1.044211, -1.063170, -1.135735, -1.081859, -1.034380]

err_aer = np.abs(np.array(E_sv) - np.array(E_aer))
err_qpu = np.abs(np.array(E_sv) - np.array(E_qpu))
err_mit = np.abs(np.array(E_sv) - np.array(E_mit))

plt.figure(figsize=(8,5))
plt.plot(bond_lengths, err_aer, 'o--', label='Aer Simulator', color='tab:blue')
plt.plot(bond_lengths, err_qpu, 's--', label='QPU (Unmitigated)', color='tab:red')
plt.plot(bond_lengths, err_mit, '^--',  label='QPU (DD+ZNE Mitigated)', color='tab:orange')
plt.axhline(y=0.0016, color='black', linestyle=':', label='Chemical Accuracy (1 kcal/mol)')
plt.xlabel('Bond Length (Å)', fontsize=12)
plt.ylabel(r'$|E − E_{statevector}|$ (Hartree)', fontsize=12)
plt.title('Absolute Energy Error vs Bond Length', fontsize=13)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('error_analysis.png', dpi=150)
plt.show()
```
