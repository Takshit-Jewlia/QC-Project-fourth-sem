```python
import pandas as pd
import pickle
import numpy as np
import matplotlib.pyplot as plt

files = ["Bond_45","Bond_60","Bond_75","Bond_90","Bond_105","Bond_120"]
df1 = pd.read_pickle("Bond_45/Result_dataframe.pkl")
df2 = pd.read_pickle("Bond_60/Result_dataframe.pkl")
df3 = pd.read_pickle("Bond_75/Result_dataframe.pkl") 
df4 = pd.read_pickle("Bond_90/Result_dataframe.pkl")
df5 = pd.read_pickle("Bond_105/Result_dataframe.pkl")
df6 = pd.read_pickle("Bond_120/Result_dataframe.pkl")

df_QPU = pd.concat([df1, df2, df3, df4, df5, df6], ignore_index=True)
df_QPU.head()
```

```python
df_Aer = pd.read_pickle("Data_results/Aer_Simulator_Results/Aer_Results_2.pkl")
df_Aer.head()
df_Svec = pd.read_pickle("Data_results/Statevector_Results/Statevector_Results.pkl")
df_Svec.head()
```

```python

plt.plot(df_Aer["Bond Length"], df_Aer["Ground State Energy (Hartree)"], marker='.', label='Aer Simulator')
plt.plot(df_QPU["Bond Length"], df_QPU["Ground State Energy (Hartree)"], marker='.', color='red', label='QPU')
plt.plot(df_Svec["Bond Length"], df_Svec["Ground State Energy (Hartree)"], marker='.', color='green', label='Statevector')
plt.xlabel("Bond Length (Å)")
plt.ylabel("Ground State Energy (Hartree)") 
plt.grid(True)
plt.title("H2 Molecule Potential Energy Curve")
plt.legend()
plt.show()
```

```python
df_Aer["Bond Length"] = df_Aer["Bond Length"].round(2)
df_Svec["Bond Length"] = df_Svec["Bond Length"].round(2)
df_QPU["Bond Length"] = df_QPU["Bond Length"].round(2)
```

```python
df_results = df_combined[["Bond Length", "Ground State Energy (Hartree)_QPU", "Ground State Energy (Hartree)_Aer", "Ground State Energy (Hartree)"]]
df_results.rename(columns={"Ground State Energy (Hartree)_QPU": "GS Energy QPU", "Ground State Energy (Hartree)_Aer": "GS Energy Aer", "Ground State Energy (Hartree)": "GS Energy Svec"}, inplace=True)
df_results.head(7)
```

```python
for history in df_combined["Convergence History_QPU"]:
    plt.plot(history, marker='.')
plt.xlabel("Iteration Number")
plt.ylabel("Energy (Hartree)")
plt.title("Convergence History of VQE on QPU")
plt.grid(True)
plt.legend([f"{r:.2f} Å" for r in df_combined["Bond Length"]], loc='upper right')
plt.show()
```

```python
min_bond_Svec = df_Svec.loc[df_Svec["Ground State Energy (Hartree)"].idxmin(), "Bond Length"]
min_bond_Aer = df_Aer.loc[df_Aer["Ground State Energy (Hartree)"].idxmin(), "Bond Length"]
min_bond_QPU = df_QPU.loc[df_QPU["Ground State Energy (Hartree)"].idxmin(), "Bond Length"]

print(f"Bond length with minimum energy (Svec): {min_bond_Svec} Angstroms")
print(f"Bond length with minimum energy (Aer): {min_bond_Aer} Angstroms")
print(f"Bond length with minimum energy (QPU): {min_bond_QPU} Angstroms")
```

```python
# Extract optimal parameters for each simulator at each bond length
optimal_params_aer = df_Aer[["Bond Length", "optimal_parameters_list"]].rename(columns={"optimal_parameters_list": "Optimal Parameters (Aer)"})
optimal_params_svec = df_Svec[["Bond Length", "optimal_parameters_list"]].rename(columns={"optimal_parameters_list": "Optimal Parameters (Svec)"})
optimal_params_qpu = df_QPU[["Bond Length", "optimal_parameters_list"]].rename(columns={"optimal_parameters_list": "Optimal Parameters (QPU)"})

# Merge on bond length
optimal_params_comparison = optimal_params_aer.merge(optimal_params_svec, on="Bond Length").merge(optimal_params_qpu, on="Bond Length")

optimal_params_comparison
```
