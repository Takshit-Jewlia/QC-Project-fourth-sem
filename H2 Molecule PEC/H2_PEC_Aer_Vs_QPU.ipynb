```python
import pandas as pd
import pickle
import numpy as np
import matplotlib.pyplot as plt

files = ["Bond_045","Bond_060","Bond_075","Bond_090","Bond_105","Bond_120"]

df1 = pd.read_pickle("Data_results/Bond_045/Result_dataframe.pkl")
df2 = pd.read_pickle("Data_results/Bond_060/Result_dataframe.pkl")
df3 = pd.read_pickle("Data_results/Bond_075/Result_dataframe.pkl") 
df4 = pd.read_pickle("Data_results/Bond_090/Result_dataframe.pkl")
df5 = pd.read_pickle("Data_results/Bond_105/Result_dataframe.pkl")
df6 = pd.read_pickle("Data_results/Bond_120/Result_dataframe.pkl")
df7 = pd.read_pickle("Data_results/Bond_135/Result_dataframe.pkl")

df = pd.concat([df1, df2, df3, df4, df5, df6, df7], ignore_index=True)
df.head()
```

```python
df_Aer = pd.read_pickle("Data_results/Aer_Simulator_Results/Aer_Results_2.pkl")
df_Aer.head()
```

```python

plt.plot(df_Aer["Bond Length"], df_Aer["Ground State Energy (Hartree)"], marker='o', label='Aer Simulator')
plt.plot(df["Bond Length"], df["Ground State Energy (Hartree)"], marker='o', color='red', label='QPU')
plt.xlabel("Bond Length (Å)")
plt.ylabel("Ground State Energy (Hartree)") 
plt.grid(True)
plt.title("H2 Molecule Potential Energy Curve")
plt.legend()
plt.show()
```

```python

df_QPU = pd.concat([df1, df2, df4, df5, df6, df7], ignore_index=True)
```

```python
df_QPU["Total Energy History"] = df_QPU.apply(
    lambda row: [e + row["Nuclear Repulsion Energy (Hartree)"]
                 for e in row["Convergence History"]],
    axis=1
)

df_QPU["Min Total Energy"] = df_QPU["Total Energy History"].apply(min)
df_QPU["Min Electronic Energy"] = df_QPU["Convergence History"].apply(min)
```
